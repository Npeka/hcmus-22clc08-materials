```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Đọc Dataframe từ file csv, với các cột như sau
field_names = ['Region', 'Country', 'HIV Treatment Rate', 
               'GNI per capita (USD)', 'GDP growth (annual %)',
               'Adult literacy rate','GDP per capita growth']
world_df = pd.read_csv('Data/world_data.csv',skiprows=1,names=field_names)
world_df
```

```python
# Vẽ barplot cho các Region

# TODO
world_df['Region'].value_counts().plot(kind='barh', width=0.85)
plt.title('Distribution of contries by the regions')
plt.xlabel('Count')
plt.ylabel('Region')
plt.show()
```

```python
# Vẽ boxplot cho 'GNI per capita (USD)'

# TODO
world_df['GNI per capita (USD)'].dropna().plot(kind='box', color='black', medianprops=dict(color='orange')) 
plt.ylim(0, 60000)
```

```python
# Vẽ violinplot cho 'GNI per capita (USD)'

# TODO
# dont show the 'bodies
plt.violinplot(world_df['GNI per capita (USD)'].dropna())
plt.show()
```

```python
# Vẽ boxplot của 'GDP growth (annual %)' thuộc các Region lên cùng một biểu đồ

# TODO

world_df.boxplot(
    column='GDP growth (annual %)', 
    by='Region', 
    grid=False, 
    color='black', 
    medianprops=dict(color='orange'),
)
plt.title('')
plt.suptitle('')
plt.xlabel('')
plt.xticks(rotation=90)
plt.ylim(-30, 10)
plt.show()
```

```python
# Vẽ histogram của 'GNI per capita (USD)'

# TODO
world_df['GNI per capita (USD)'].dropna().plot(
    kind='hist', 
    bins=20, 
    edgecolor = 'white',
)
plt.title('Distribution of GNI per capita')
plt.xlabel('GNI per capita')
plt.ylabel('Density')
plt.show()
```

Gọi các vùng sau là bán cầu bắc (Northern)
- 'Europe & Central Asia'
- 'North America'
- 'East Asia & Pacific'
  
Gọi các vùng sau là bán cầu nam (Southern)
- 'Sub-Saharan Africa'
- 'Middle East & North Africa'
- 'South Asia'
- 'Latin America & Caribbean'

```python
# Hãy vẽ 2 histogram thể hiện sự phân bố của cột 'GNI per capita (USD)'
# cho bán cầu bắc và bán cầu nam lên cùng một bản đồ

# TODO
northern_regions = ['Europe & Central Asia', 'North America', 'East Asia & Pacific']
southern_regions = ['Sub-Saharan Africa', 'Middle East & North Africa', 'South Asia', 'Latin America & Caribbean']

northern_df = world_df[world_df['Region'].isin(northern_regions)]
southern_df = world_df[world_df['Region'].isin(southern_regions)]

plt.hist(northern_df['GNI per capita (USD)'].dropna(), bins=20, alpha=0.25, label='Northern', edgecolor='black', density=True)
plt.hist(southern_df['GNI per capita (USD)'].dropna(), bins=20, color='orange', alpha=0.25, label='Southern', edgecolor='black', density=True)
plt.title('Distribution of gross national income per capita')
plt.xlabel('GNI per capita')
plt.ylabel('Density')
plt.grid(False)
plt.legend()
plt.show()
```

```python
# Hãy vẽ scatter plot thể hiện mối quan hệ giữa
#'GNI per capita (USD)' và 'Adult literacy rate'

# TODO
world_df.plot(
    kind='scatter',
    x='GNI per capita (USD)',
    y='Adult literacy rate',
)
```

```python
# Hãy vẽ scatter plot thể hiện mối quan hệ giữa
#'GNI per capita (USD)' và 'Adult literacy rate'
# cho mỗi Region với màu khác nhau
# TODO
regions_colors = {
    'Europe & Central Asia': '#1F77B4',
    'Sub-Saharan Africa': 'orange',
    'Latin America & Caribbean': 'green',
    'East Asia & Pacific': 'red',
    'Middle East & North Africa': 'purple',
    'South Asia': 'brown',
    'North America': 'pink',
}
for region, color in regions_colors.items():
    region_df = world_df[world_df['Region'] == region]
    plt.scatter(
        region_df['GNI per capita (USD)'],
        region_df['Adult literacy rate'],
        color=color,
        label=region,
    )
plt.xlabel('GNI per capita (USD)')
plt.ylabel('Adult literacy rate')
plt.legend()
plt.show()
```

```python
# Hãy vẽ scatter plot thể hiện mối quan hệ giữa
#'GNI per capita (USD)' và 'Adult literacy rate'
# Cùng với hai histogram ở hai trục

# TODO
fig = plt.figure(figsize=(6, 6))
grid = plt.GridSpec(6, 6, hspace=0.2, wspace=0.2)
main_ax = fig.add_subplot(grid[2:, :-1])
x_hist = fig.add_subplot(grid[1, :-1], sharex=main_ax)
y_hist = fig.add_subplot(grid[2:, -1], sharey=main_ax)

x, y = world_df['GNI per capita (USD)'], world_df['Adult literacy rate']

main_ax.set_xlabel('GNI per capita (USD)')
main_ax.set_ylabel('Adult literacy rate')
main_ax.plot(x, y, 'o', markersize=5, alpha=0.5)

x_hist.hist(x, bins=15, color='blue', orientation='vertical', edgecolor='black')
y_hist.hist(y, bins=15, color='red', orientation='horizontal', edgecolor='black')

x_hist.tick_params(axis='x', labelbottom=False) 
y_hist.tick_params(axis='y', labelleft=False)

plt.show()
```
