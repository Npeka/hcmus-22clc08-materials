Đúng như dự đoán, tất cả các cột đều đang có kiểu dữ liệu là str. Để có thể tiếp tục khám phá tiếp thì ta cần tiền xử lý để các cột có kiểu dữ liệu phù hợp.

- Cột "State": string là đúng rồi, không cần chuyển
- Cột "Started on" và "Completed on": chuyển sang kiểu dữ liệu datetime
- Cột "Time taken": chuyển sang một con số thể hiện cho số phút
- Các cột "Grade/16.00", "Q. 1 /1.00", ...: chuyển sang kiểu dữ liệu số

## Tiền xử lý

```python
new_cols = {}
for col_name, col_vals in cols.items():
    if '/' in col_name:
        new_col_vals = []
        for col_val in col_vals:
            try:
                new_col_vals.append(float(col_val))
            except:
                new_col_vals.append(0.0)
        new_cols[col_name] = new_col_vals
    elif col_name in ['Started on', 'Completed']:
        new_col_vals = []
        for col_val in col_vals:
            new_col_val = dt.datetime.strptime(col_val, '%d/%m/%Y %H:%M')
            new_col_vals.append(new_col_val)
        new_cols[col_name] = new_col_vals
    elif col_name == 'Time taken':
        new_col_vals = []
        for col_val in col_vals:
            hour = 0; minute = 0; second = 0
            temp_list = col_val.split(' ')
            for temp_i in range(len(temp_list)):
                if 'hour' in temp_list[temp_i]:
                    hour = int(temp_list[temp_i - 1])
                elif 'min' in temp_list[temp_i]:
                    minute = int(temp_list[temp_i - 1])
                elif 'sec' in temp_list[temp_i]:
                    second = int(temp_list[temp_i - 1])
            new_col_val = hour * 60 + minute + second / 60
            new_col_vals.append(new_col_val)
        new_cols[col_name] = new_col_vals
    else:
        new_cols[col_name] = col_vals
```

## Tiền xử lý + phân tích dữ liệu để trả lời cho câu hỏi 1

*Câu hỏi 1: Cột điểm tổng “Grade/10.00” có phân bố các giá trị như thế nào?*

Ở đây không cần tiền xử lý mà sẽ đi luôn vào phân tích dữ liệu.

Đầu tiên, ta sẽ xem thử median của cột "Grade/10.00".

```python
grades = cols['Grade/10.00']
sorted_grades = sorted(grades)
p = 50
i = int(p/100 * (len(sorted_grades)-1))
sorted_grades[i]
```

Để cảm thụ rõ hơn về phân bố các giá trị của cột "Grade/10.00", ta sẽ chia miền giá trị của cột này thành các khoảng (bin) và xem thử mỗi khoảng có bao nhiêu sinh viên.

Giả sử ta sẽ chia đoạn [0, 10] thành 5 bin như sau: [0, 2], (2, 4], (4, 6], (6, 8], (8, 10]. Ta có thể lưu kết quả vào một dictionary, trong đó: key là một tuple cho biết 2 đầu của một bin (thay tuple bằng list thì có được không?), value là số lượng sinh viên rơi vào bin này.

***Sinh viên hoàn thành tiếp phần này như bài tập về nhà***

```python
bin_ranges = [(0, 2), (2, 4), (4, 6), (6, 8), (8, 10)]
bin_counts = {bin_range: 0 for bin_range in bin_ranges}

grades = cols['Grade/10.00']
for grade in grades:
    for bin_range in bin_ranges:
        low, high = bin_range
        if (low == 0 and grade == low) or low < grade <= high:
            bin_counts[bin_range] += 1
            break

bin_counts
```

**Trả lời câu hỏi 1.1:** *Cột điểm tổng “Grade/10.00” có phân bố các giá trị như thế nào?*

Kết quả thống kê cho thấy số lượng sinh viên trong mỗi bin như sau:
- Từ 0 đến 2: 0 sinh viên
- Từ 2 đến 4: 1 sinh viên
- Từ 4 đến 6: 9 sinh viên
- Từ 6 đến 8: 9 sinh viên
- Từ 8 đến 10: 40 sinh viên

Kết quả trên cho thấy hầu hết sinh viên có điểm từ 8 đến 10, nghĩa là phần lớn điểm tập trung ở mức cao, qua đó có thể đánh giá tích cực về kết quả của bài kiểm tra.

**Trả lời câu hỏi 1.2:** *Thay tuple bằng list thì có được không?*

Ta không thể sử dụng `list` thay `tuple` để làm `key` trong `dictionary` với những lý do sau đây: 
- `key` phải là kiểu dữ liệu bất biến `immutable` và có thể hash `hashable`. Do đó ta cần chọn một kiểu dữ liệu `immutable` tương ứng để làm `key`.
- Theo tính chất của `key`, ta không thể chọn `list` vì list là kiểu dữ liệu có thể thay đổi `mutable` và không thể hash `unhashable`.
- `tuple` là một kiểu dữ bất biến `immutable` và có thể hash `hashable` đáp ứng yêu cầu của `key`.

Vì vậy, ta không thể thay `tuple` bằng `list`. Sử dụng `tuple` làm `key` trong `dictionary` sẽ đảm bảo rằng giá trị của `key` là bất biến, một `key` sẽ được hash thành một mã duy nhất không thể thay đổi. Điều này giúp dữ liệu đáng tin cậy và an toàn hơn.

Tuy không thể sử dụng trực tiếp `list` để làm `key`, ta vẫn có thể sử dụng gián tiếp bằng cách chuyển `list` sang `string`, `frozenset` hoặc một kiểu dữ liệu `immutable` khác. Nhưng cũng cần lưu ý rằng việc này có thể làm phức tạp đoạn mã và khó quản lý hơn.

## Tiền xử lý + phân tích dữ liệu để trả lời cho câu hỏi 2

*Câu hỏi 2: Câu quiz nào có nhiều bạn làm sai nhất, câu nào có nhiều bạn làm sai nhì, ...?*

Ta có thể lưu kết quả vào một dictionary, trong đó: key là tên của cột ứng với câu quiz, value là điểm trung bình của cột. Dictionary này được sort tăng dần theo value.

***Sinh viên hoàn thành tiếp phần này như bài tập về nhà***

```python
fail_counts = {}
average_grades = {}

for question, grades in cols.items():
    if 'Q.' in question:
        fail_count = 0
        total_score = 0
        for grade in grades:
            total_score += grade
            if grade < 0.53:
                fail_count += 1
        fail_counts[question] = fail_count
        average_grades[question] = total_score / len(grades)
```

```python
# Sắp xếp theo số câu sai giảm dần để tương ứng với số điểm trung bình tăng dần
sorted_fail_counts = sorted(fail_counts.items(), key=lambda item: item[1], reverse=True)

print('Câu hỏi \t Số câu sai \n' + '-'*27)
for question, count in sorted_fail_counts:
    print(f'{question} \t {count}')
```

```python
# điểm trung bình được sắp xếp tăng dần
sorted_average_grades = sorted(average_grades.items(), key=lambda item: item[1])

print('Câu hỏi \t Điểm trung bình \n' + '-'*36)
for question, avg in sorted_average_grades:
    print(f'{question} \t {avg}')
```
